```python
import pandas as pd
import sys
import numpy as np

sys.path.append('../../scripts/3_signal_analysis')

from calculate import get_signals
from save import save_signals
```

### **a) DMSO/IAA signals:** Calculate signals from .wig file from provided coordinates in .bed format

```python
#################### EXAMPLE USAGE ####################
################ PASTE YOUR PATHS INSTEAD #############
output_dir = '../../results/signals_spikein/RDY374'

# When comparing signals from different conditions, use same coordinates
# In case DMSOvsIAA, it's common peaks from compare_peaks.ipynb
bed_path = '../../results/compared_peaks/RDY374_as_DMSO_25_nC_RDY374_as_IAA_25_nC/in_RDY374_as_IAA_25_nC_RDY374_as_DMSO_25_nC_ex.bed'

DMSO_wig_path = '../../data/wig/spikein_mean/RDY374_as_DMSO_25_nC_spikein_mean.wig'
IAA_wig_path = '../../data/wig/spikein_mean/RDY374_as_IAA_25_nC_spikein_mean.wig'

# Read files
coord = pd.read_csv(bed_path, names=['chr', 'start', 'end', 'center'], sep='\t', header=None)
DMSO_wig = pd.read_csv(DMSO_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
IAA_wig = pd.read_csv(IAA_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
```

```python
DMSO_signals = get_signals(coord, DMSO_wig)
IAA_signals = get_signals(coord, IAA_wig)

# Save signals
save_signals(signals=DMSO_signals,
            in_file_path=DMSO_wig_path,
            out_dir_path=output_dir)
save_signals(signals=IAA_signals,
            in_file_path=IAA_wig_path,
            out_dir_path=output_dir)

# Clean cache, otherwise will crash
del coord, DMSO_wig, IAA_wig
```

### **c) Signals from 4 different stages of cell cycle - asyn, G1, S, G2/M** - cpm signals

```python
#################### EXAMPLE USAGE ####################
################ PASTE YOUR PATHS INSTEAD #############
output_dir = '../results/signals_cpm/RDY412'

# When comparing signals from different conditions, use same coordinates
# In case phases, it's common peaks from all phases from compare_peaks.ipynb
bed_path = '../results/compared_peaks/RDY..'

as_wig_path = '../../data/wig/cpm_mean/RDY412_as_unTr_nT_nC_cpm_mean.wig'
G1_wig_path = '../../data/wig/cpm_mean/RDY412_G1_unTr_nT_nC_cpm_mean.wig'
S_wig_path = '../../data/wig/cpm_mean/RDY412_S_unTr_nT_nC_cpm_mean.wig'
G2_wig_path = '../../data/wig/cpm_mean/RDY412_G2_unTr_nT_nC_cpm_mean.wig'

# Read files
coord = pd.read_csv(bed_path, names=['chr', 'start', 'end', 'center'], sep='\t', header=None)

as_wig = pd.read_csv(as_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
G1_wig = pd.read_csv(G1_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
S_wig = pd.read_csv(S_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
G2_wig = pd.read_csv(G2_wig_path, names=['chr', 'start', 'end', 'signal'], sep='\t', header=None, comment='#')
```

```python
as_signals = get_signals(coord, as_wig)
G1_signals = get_signals(coord, G1_wig)
S_signals = get_signals(coord, S_wig)
G2_signals = get_signals(coord, G2_wig)

# Save signals
save_signals(signals=as_signals,
            in_file_path=as_wig_path,
            out_dir_path=output_dir)

save_signals(signals=G1_signals,
            in_file_path=G1_wig_path,
            out_dir_path=output_dir)

save_signals(signals=S_signals,
            in_file_path=S_wig_path,
            out_dir_path=output_dir)

save_signals(signals=G2_signals,
            in_file_path=G2_wig_path,
            out_dir_path=output_dir)

# Clean cache, otherwise will crash
del coord, as_wig, G1_wig, S_wig, G2_wig
```
